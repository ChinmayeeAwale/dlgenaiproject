Prepare local model artifacts for Ollama serving.

```python
from pathlib import Path\n
modelfile = Path('../models/gguf/Modelfile')\n
template = '''FROM ./finance-rag.gguf\n\nSYSTEM You are a finance assistant. Use retrieved context and cite sources.\n\nPARAMETER temperature 0.1\n'''\n
modelfile.parent.mkdir(parents=True, exist_ok=True)\n
modelfile.write_text(template, encoding='utf-8')\n
print('Wrote', modelfile)
```
